# Your data analysis code 

```python
import pandas as pd
```

```python
df = pd.read_csv('german_credit_data_processed.csv')
print(df.shape)
df.head()

# The dataset has 1000 rows and 21 columns. Out of 21 columns 20 of them are features and target is 'goodness'.
```

```python
df.dtypes

# There are 7 numerical and 13 categorical features, which matches the provided documentation.
```

```python
df.isnull().sum()

# checking for missing values - No missing values
```

```python
num_cols = ['duration in month', 
            'credit amount', 
            'installment rate',
            'residence since', 
            'relationship : age', 
            'existing credits', 
            'people to provide maintenance for']

df[num_cols].describe()

# Average loan duration is around 21 months, but it can go up to 72 months
# Ages range from 19 to 75, with an average of around 35 years.
```

```python
print(df['goodness'].value_counts())

# 70% of applicants are classified as Good (1) and 30% as Bad (2). This is an imbalanced dataset.
```

```python
df['checking account'].value_counts()

# Almost 40% of applicants have no checking account at all, yet they still applied for credit.
```

```python
df['credit history'].value_counts()

# More than half have existing credits that are being paid on time.
```

```python
df['credit purpose'].value_counts()

# Top 3 purposes are radio/TV, new car, and furniture
```

```python
df['savings account'].value_counts()

# Over 60% of applicants have less than 100 DM (Deutsche Mark) in savings and  18% have no savings account at all.
```

```python
df['employment since..'].value_counts()

# Most applicants are employed. Only 6.2% are unemployed. The largest group is the people with 1–4 years of employment.
```

```python
df['status : sex'].value_counts()

# There is no 'female : single' category at all in this dataset.
```

```python
df['housing'].value_counts()

# 71.3% own their home which may contradict with low savings balance.
```

```python
df['foreign worker'].value_counts()

# 96.3% of applicants are foreign workers which is the most unusual finding in the dataset.
```

- *Who created the dataset?* The dataset was created by the CDC (Centers for Disease Control and Prevention) through their BRFSS (Behavioral Risk Factor Surveillance System) survey.

- *For what purpose was the dataset created?* To better understand the relationship between  lifestyle and diabetes in the US.

- *Who funded the dataset creation?* The CDC.

- *How was the dataset collected?* The dataset is built from a combined landline and cell phone survey. It includes data from all 50 states.

- *How is the data sampled? Does the sampling make sense?* The 2014 BRFSS data reflects changes initially made in 2011 in weighting methodology (Raking) and the addition of cell phone only respondents. 'Raking' is a statistical technique to make the sample more representative of the actual population.

- *Does the dataset make any assumptions?* It assumes self-reported answers are accurate.

- *Were there data points that have been specifically left out?*  Some things are likely excluded - like race, geographic identifiers etc. even though the original BRFSS collects it.

- *Are there any known errors or inconsistencies in the dataset?* The BRFSS 2014 data is not directly comparable to years of BRFSS data before 2011 because of the changes in weighting methodology and the addition of the cell phone sampling frame.

- *Are there missing values?* No.

- *Is the meaning behind every attribute in the dataset clear?* Mostly Yes.

- *Does the dataset contain metadata, e.g., a README file?* The CDC provides a full BRFSS Overview PDF covering background, data collection, processing, and statistical/analytical issues.
