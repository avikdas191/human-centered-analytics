# Group Project Proposal

The goal of the group project is to create a comprehensive explanation interface to help a target audience make informed decisions, in which we apply the techniques and concepts learned throughout the semester.

Deliverables of the group project (at the end of the semester):
1. Explanation interface (the code in your repo)
2. Project description (The README.md in your repo, will constantly be updated)
3. A 10 minute project presentation that includes all information from the project description (last week of the semester)

Criteria for a successful group project:
- Technical implementation: use of the learned techniques, readability of your code
and documentation
- Presentation: engaging, easy to follow presentation that provides insights into your
project process and results (e.g., reasoning for design decisions, interview outcomes)
- Information Design: Organization, logical flow, grouping, and hierarchy of elements
in the interface.
- Creativity and Innovation: Creative and innovative approaches in presenting the
data, unique visual metaphors, storytelling techniques, or interactive features.
- Audience: Suitability of the design choices to engage and resonate
with the intended audience (target audience / user), ability to capture attention and interest.
- Insightfulness: Ability to uncover meaningful insights or patterns in the data,
revealing trends, relationships, or disparities.

You have already formed groups, selected your dataset(s) in the health/medical domain, and formulated initial questions worth answering. This assignment will take the project description further by defining a concrete, realistic decision-making scenario in which a machine learning model will be applied based on your dataset.

**✏️ Task** — **Write a project proposal and document in your repository README.md**

By now, your README.md already includes information about your chosen dataset and some questions you might want to answer with the dataset. The goal of this assignment is to advance the README.md file with an initial project proposal that:

- Clearly describes the dataset(s) you will use
- Defines a real-world decision-making scenario relevant to the dataset
- List and describe potential stakeholders
