# Group Project I — Data Analysis & Formulating Questions

## Introduction
The goal of the HCDS group project is to create a comprehensive explanation interface to help a target audience make informed decisions in an AI assisted decision making scenario, which you will present in the last lecture of the semester. More detailed instructions will be provided in the following weekly assignments. This week we will begin with some preparations for the group projects.

**✏️ Task 1** — **Putting the team together**  
As the name implies, this project will be a group effort.
- Get together into groups of 5 members.

**Note:** *If you are in a group of less than 5, please join with another group. If you have trouble finding a group, please post a message in the Whiteboard Forum. In any case, please be proactive if you are having troubles finding a group.*


**✏️ Task 2** — **Dataset selection & Brain Storming**  
You will need a dataset to work with in your group project, and you can choose it yourself.

Some restriction for the dataset:
- It must be part of the health/medical domain
- You should be able to train a model on it (We recommend tabular data)
- It should make sense to evaluate fairness metrics on the data.
- The dataset should be publicly available and you should have the rights to use it.

Steps for your group to take:
- Select a dataset to work with for the rest of the project given the restrictions above. You can also choose multiple datasets and merge them.
- Find and document all relevant information for this dataset. Explore the questions we answered in previous assignments.
- Prepare a set of possible questions that you could answer with this data in your group project (see the first step of the data science lifecycle). The model you will train should be suitable to help answering your questions.

**Note:** *In the Resources section below, we give you sources for finding datasets and some examples of datasets you can use.*
